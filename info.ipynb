Conclusión General
Desde la perspectiva del diseño racional de fármacos, la respuesta a la pregunta ¿Son similares los modos de unión y las interacciones con la proteína? es un rotundo NO (a excepción del control de redocking del propio 7IK).

Sobre los candidatos (DB06836 y DB11962): Ambos fármacos se unen en la región general del sitio activo, pero fracasan en mimetizar el modo de anclaje del inhibidor nativo. Sus estructuras moleculares (muy extendida en uno, muy rígida y voluminosa en el otro) les impiden adoptar la geometría necesaria para solaparse con el 7IK.
Implicación Biológica: Esto no significa necesariamente que no funcionen, pero indica que su mecanismo de inhibición dependería de crear un patrón de interacciones completamente nuevo y alternativo al del ligando natural. No actúan como "análogos estructurales" perfectos del 7IK, sino como ocupantes estéricos que bloquean el sitio de una forma geométrica distinta.

# Punto 4: Análisis de Resultados del Cribado Virtual (Screening)

Tras ejecutar el cribado virtual masivo (Virtual Screening) utilizando AutoDock Vina sobre una librería de 20 ligandos frente a dos receptores distintos (Proteína Humana y Proteína Fúngica), se obtuvieron los histogramas comparativos y las tablas de afinidad de unión ($\Delta G$ en kcal/mol).

A continuación se presenta el análisis de dichos resultados:

<img src="analysis/virtual_screening/screening/histograma_comparativo.png">

### 1. ¿Tienen sentido los resultados y qué representan?
Sí, los resultados **tienen mucho sentido** en el contexto de diseño y reposicionamiento de fármacos. 
*   **El valor de $\Delta G$ (Energía Libre de Gibbs):** Representa la afinidad termodinámica pronosticada. Valores más negativos indican una interacción más fuerte y espontánea entre el ligando y la proteína (mayor probabilidad de inhibición efectiva).
*   **La línea punteada roja (-7.0 kcal/mol):** Actúa como un umbral (cutoff) comúnmente aceptado en simulaciones *in silico* para definir qué compuestos se consideran "hits" (aciertos) prometedores que merecen mayor investigación.

### 2. Comparación: Proteína Humana vs. Proteína Fúngica

Al observar los histogramas y los datos estadísticos, surgen diferencias claras en el perfil de unión de ambas especies:

*   **Afinidad Global y Distribución:**
    *   **Proteína Humana:** Muestra una curva de distribución normal más estrecha y desplazada hacia valores más negativos (Mediana = -8.15 kcal/mol). **El 100% de los ligandos (20/20)** superó el umbral de -7.0 kcal/mol. Esto indica que el bolsillo activo humano de esta enzima es sumamente permisivo y afín a esta quimioteca en particular.
    *   **Proteína Fúngica:** La distribución es más ancha y está desplazada a la derecha, indicando menor afinidad promedio (Mediana = -7.15 kcal/mol). Solo el **65% de los ligandos (13/20)** se consideraron "hits". El sitio activo fúngico parece ser más restrictivo o diferente topológicamente, generando choques estéricos o peores contactos para varios de los compuestos probados.

*   **Identificación de Candidatos y Selectividad (Análisis de los Scores):**
    El objetivo ideal al buscar un fármaco antifúngico es encontrar un compuesto que inhiba potentemente la enzima del hongo, pero que tenga menor o igual efecto en la humana para reducir la toxicidad. Al cruzar los datos de los archivos `.csv`, descubrimos perfiles muy interesantes:
    
    1.  **Candidatos Selectivos para el Hongo (Los mejores prospectos):**
        *   **35642:** Humana (-8.4 kcal/mol) vs **Hongo (-9.0 kcal/mol)**. Es más afín al hongo.
        *   **HY-144894:** Humana (-8.1 kcal/mol) vs **Hongo (-9.0 kcal/mol)**. Significativamente mejor para el hongo.
        *   **HY-135805A:** Humana (-8.9 kcal/mol) vs **Hongo (-9.0 kcal/mol)**. Alta potencia en ambos, leve preferencia por el hongo.
    
    2.  **Candidatos Altamente Tóxicos (Selectivos para Humana):**
        *   **HY-14881:** Es el mejor inhibidor para la proteína humana (**-9.2 kcal/mol**), pero en el hongo tiene un score muy pobre (-6.7 kcal/mol, ni siquiera es hit). Este compuesto es el ejemplo perfecto de algo que causaría muchos efectos secundarios sin matar al hongo eficientemente.

### Conclusión del Screening
El análisis comparativo es concluyente: **las diferencias estructurales entre la enzima humana y la fúngica dictan perfiles de afinidad distintos**. Aunque la proteína humana es en general más susceptible a estos ligandos, el cribado virtual cumplió su propósito aislando compuestos específicos (como el **35642** y el **HY-144894**) que logran un anclaje superior en la variante del hongo (-9.0 kcal/mol). Estos deberían ser los compuestos priorizados para fases de validación experimental o *Molecular Dynamics*.
